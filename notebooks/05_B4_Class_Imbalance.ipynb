# 05 · B4 — Which imbalance strategy helps rare attacks without raising false alarms?
Top-4 models from B2 (chosen on **validation** macro-F1) × {none, weighted CE, focal (γ = 2), logit-adjusted loss, class-balanced sampling, SMOTE (tabular only)}. For tree models, "weighted CE / focal / logit-adjusted / balanced sampling" are all implemented as inverse-frequency sample weights (their tree analogue); report them once as "class weighting".

Outcome: the minority-recall versus false-positive-rate trade-off.

```python
import sys, os, json, warnings
sys.path.insert(0, os.path.abspath('..'))
warnings.filterwarnings('ignore')
import numpy as np, pandas as pd
pd.set_option('display.width', 200, 'display.max_columns', 60)
from sdnids_bench.config import load_config, mode_params, active_datasets, enable_determinism
cfg = load_config('..')
mp = mode_params(cfg)
enable_determinism()
print('mode:', cfg['mode'], '| results:', cfg['out_dir'])
from sdnids_bench.experiments import run_grid, load_results, choose_best_L, choose_top_models
DATASETS = active_datasets(cfg, 'B4')
best_L = choose_best_L(cfg)
models = choose_top_models(cfg, k=4)
print('B4 on:', DATASETS, '| models:', models, '| L:', best_L)
```

```python
run_grid(cfg, 'B4', DATASETS, ['multi'], ['SC'], [best_L], mp['sweep_seeds'], models, imbalances=cfg['b4_strategies'])
```

```python
res = load_results(cfg, DATASETS)
b4 = res[res.exp=='B4']
display(b4.groupby(['dataset','model','imbalance'])[['macro_f1','minority_recall','fpr']].mean().round(4))
```
